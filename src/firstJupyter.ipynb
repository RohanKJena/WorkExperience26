### Bionomial Theorem: $$(a+b)^n = \sum_{k=0}^{n} \binom{n}{k} a^{n-k}b^k$$


```python
# Enable inline plotting in Jupyter
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.linspace(0, 10, 100)
y = np.tan(x)

# Create the plot
plt.figure(figsize=(8, 4))
plt.plot(x, y, label='Tan Wave', color='red', linewidth=2)

# Add titles and labels
plt.title('Sine Wave Plot')
plt.xlabel('X-Axis (Values)')
plt.ylabel('Y-Axis (Sin(x))')

# Add legend and grid
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
```
